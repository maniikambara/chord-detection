```python
import numpy as np
import pandas as pd
import librosa as lb
from IPython.display import display, Audio
import sklearn as sl
from pathlib import Path
```

```python
# ---- Data ----
DATASET_DIRS = [
    "guitar_chords_v3",
    "piano_triads"
]

FORMAT = ".wav"
```

```python
def find_data(folder: Path):
    for data in folder.rglob("*"):
        if data.is_file() and data.suffix.lower() == FORMAT:
            yield data

def build_index(data_dir: dict) ->pd.DataFrame:
    roots = {
        Path(value) for value in data_dir
    }
    for root in roots:
        if not root.exists():
            raise FileNotFoundError(
                f"DATA_DIR '{data_dir}' doesn't exist."
                f"Upload FakeMusicCaps to Drive and update DATA_DIR."
            )
    
    rows = []

    for dataset in roots:
        chord_folders = dataset/"chords"
        if chord_folders.exists():
            for chord in chord_folders.iterdir():
                for data in find_data(chord):
                    rows.append({"filepath": str(data),
                                 "label": chord,
                                 "source": dataset
                                 })
    
    df = pd.DataFrame(rows)
    if df.empty:
        raise RuntimeError("No audio files found. Please check DATASET_DIRS.")
    
    return df

df = build_index(DATASET_DIRS)
print(f"Found {len(df)} audio files.")
print(f"\nCounts by source:\n {df["source"].value_counts()}")
print(f"\nCounts by label:\n {df["label"].value_counts()}")
```
